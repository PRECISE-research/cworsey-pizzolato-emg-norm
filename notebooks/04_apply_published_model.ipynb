# Apply the published linear model

```python
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
TRIAL = ROOT / 'examples/full_trial_data/C06/trial_ffc10db23f/emgBPF.mot'
```

```python
from emg_normalisation import load_published_model, predict_trial, read_emg_mot
from emg_normalisation.calibration import removeGain
frame = removeGain(read_emg_mot(TRIAL), amplifier_gain=1000)
model = load_published_model(ROOT / 'models/cycling_linear/model.json')
result = predict_trial(frame, model, output_mode='scaled_envelope')
result.output.head()
```
