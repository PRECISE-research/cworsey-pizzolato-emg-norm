# Compare model outputs with the reference envelope

```python
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
TRIAL_ROOT = ROOT / 'examples/full_trial_data/C06/trial_ffc10db23f'
TRIAL = TRIAL_ROOT / 'emgBPF.mot'
REFERENCE = TRIAL_ROOT / 'reference_normalised_envelope.parquet'
```

```python
import matplotlib.pyplot as plt
from emg_normalisation import load_published_model, predict_trial, read_emg_mot
from emg_normalisation.calibration import removeGain
from emg_normalisation.io import read_table
frame = removeGain(read_emg_mot(TRIAL), amplifier_gain=1000)
reference = read_table(REFERENCE)
model = load_published_model(ROOT / 'models/cycling_linear/model.json')
scaled = predict_trial(frame, model, 'scaled_envelope').output
direct = predict_trial(frame, model, 'model_trace').output
plt.figure(figsize=(9, 3.5))
plt.plot(scaled.time, scaled.r_vaslat, alpha=0.8, label='Scaled envelope')
plt.plot(direct.time, direct.r_vaslat, linestyle='--', alpha=0.8, label='Direct model trace')
plt.plot(reference.time, reference.r_vaslat, color='black', linewidth=1.2, zorder=3, label='Reference envelope')
plt.xlabel('Time (s)'); plt.ylabel('Normalised excitation'); plt.ylim(0, 1)
plt.legend(frameon=False); plt.show()
```
